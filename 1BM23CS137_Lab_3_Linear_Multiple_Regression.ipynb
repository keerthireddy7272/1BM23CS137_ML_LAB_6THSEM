```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load dataset
df = pd.read_csv('canada_per_capita_income.csv')

# Train model
reg = LinearRegression()
reg.fit(df[['year']], df['per capita income (US$)'])

# Predict for 2020
pred_2020 = reg.predict([[2020]])
print("Predicted Income for 2020:", pred_2020[0])

# Visualization
plt.scatter(df.year, df['per capita income (US$)'], color='red', marker='+')
plt.plot(df.year, reg.predict(df[['year']]), color='blue')
plt.xlabel("Year")
plt.ylabel("Income (US$)")
plt.title("Canada Per Capita Income")
plt.show()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Load dataset
df = pd.read_csv('salary.csv')

# Preprocessing: Handle missing values in YearsExperience
df['YearsExperience'] = df['YearsExperience'].fillna(df['YearsExperience'].median())

# Train model
reg = LinearRegression()
reg.fit(df[['YearsExperience']], df['Salary'])

# Predict for 12 years of experience
pred_salary = reg.predict([[12]])
print("Predicted Salary for 12 Years Experience:", pred_salary[0])
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Load dataset
df = pd.read_csv('homeprices_Multiple_LR.csv')

# Preprocessing: Fill missing values in bedrooms
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())

# Train model
reg = LinearRegression()
reg.fit(df[['area', 'bedrooms', 'age']], df['price'])

# Example prediction: 3000 sqft, 3 bed, 40 years old
print("Predicted Home Price:", reg.predict([[3000, 3, 40]])[0])
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Load dataset
df = pd.read_csv('hiring.csv')

# Preprocessing: Fill missing values and convert words to numbers
word_to_num = {'two': 2, 'three': 3, 'five': 5, 'seven': 7, 'eight': 8, 'ten': 10, 'eleven': 11}
df['experience'] = df['experience'].fillna('zero').apply(lambda x: word_to_num.get(x, 0))
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(df['test_score(out of 10)'].median())

# Train model
reg = LinearRegression()
reg.fit(df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']], df['salary($)'])

# Predict for 12 years experience, 10 test, 10 interview
ans = reg.predict([[12, 10, 10]])
print("Predicted Salary (12, 10, 10):", ans[0])
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder

# Load dataset
df = pd.read_csv('1000_Companies.csv')

# Preprocessing: Encode categorical variable 'State'
le = LabelEncoder()
df['State'] = le.fit_transform(df['State'])

# Train model
X = df[['R&D Spend', 'Administration', 'Marketing Spend', 'State']]
y = df['Profit']
reg = LinearRegression()
reg.fit(X, y)

# Predict for specific values (Florida is encoded as a number)
florida_val = le.transform(['Florida'])[0]
pred_profit = reg.predict([[91694.48, 515841.3, 11931.24, florida_val]])
print("Predicted Profit for firm in Florida:", pred_profit[0])
```
